# **JACCARD GCG MATRIX**

```python
# cmap="YlGnBu",

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

J_gemma9b = pd.read_csv('../results/jaccard/gcg/gemma-9b/jaccard_matrix_top135_mean.csv', index_col=0)
J_llama8b = pd.read_csv('../results/jaccard/gcg/llama-8b/jaccard_matrix_top79_mean.csv', index_col=0)
J_mistral7b = pd.read_csv('../results/jaccard/gcg/mistral-7b/jaccard_matrix_top78_mean.csv', index_col=0)

# ---------------- PLOT 
def plot_jaccard(df: pd.DataFrame, title: str, figsize=(10, 9), annot=False, savepath=None):
    sns.set_theme(style="white", context="paper")
    cmap = sns.blend_palette(["#dfe1ad", "#064569",], as_cmap=True) # 

    plt.figure(figsize=figsize)
    ax = sns.heatmap(
        df, vmin=0, vmax=1, cmap=cmap, annot=True,
        square=True,
        linewidths=0.5, linecolor="white",
        cbar_kws={"label": "Jaccard similarity"}
    )
    ax.set_title(title)
    ax.set_xlabel("")
    ax.set_ylabel("")
    plt.xticks(rotation=90, ha="center")
    plt.yticks(rotation=0)
    plt.tight_layout()
    if savepath:
        plt.savefig(savepath, dpi=300, bbox_inches="tight")
    plt.show()

plot_jaccard(J_gemma9b,  "Jaccard proximity (Gemma 9B)")
plot_jaccard(J_llama8b,  "Jaccard proximity (Llama 8B)")
plot_jaccard(J_mistral7b,"Jaccard proximity (Mistral 7B)")
```
